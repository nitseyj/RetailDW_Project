# [etl].[usp_Load_DimDate]

```sql
/****** Object:  StoredProcedure [etl].[usp_Load_DimDate]    Script Date: 7/8/2026 9:44:10 PM ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

CREATE   PROCEDURE [etl].[usp_Load_DimDate]
AS
BEGIN
    SET NOCOUNT ON;

    TRUNCATE TABLE gold.DimDate;

    INSERT INTO gold.DimDate
    (
        DateKey,
        FullDate,
        CalendarYear,
        CalendarMonth,
        MonthName,
        CalendarQuarter
    )
    SELECT DISTINCT
        YEAR(OrderDate) * 10000
            + MONTH(OrderDate) * 100
            + DAY(OrderDate) AS DateKey,
        CAST(OrderDate AS DATE) AS FullDate,
        YEAR(OrderDate) AS CalendarYear,
        MONTH(OrderDate) AS CalendarMonth,
        DATENAME(MONTH, OrderDate) AS MonthName,
        DATEPART(QUARTER, OrderDate) AS CalendarQuarter
    FROM silver.Sales_Silver
    WHERE OrderDate IS NOT NULL
      AND OrderDate > '1900-01-01';
END;
GO
```

# [etl].[usp_Load_FactSales]

```sql
/****** Object:  StoredProcedure [etl].[usp_Load_FactSales]    Script Date: 7/8/2026 9:44:10 PM ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

CREATE   PROCEDURE [etl].[usp_Load_FactSales]
AS
BEGIN
    SET NOCOUNT ON;

    TRUNCATE TABLE gold.FactSales;

    INSERT INTO gold.FactSales
    (
        ProductKey,
        CustomerKey,
        DateKey,
        OrderID,
        Quantity,
        DiscountPct,
        SalesAmount,
        Profit,
        ShippingCost,
        PaymentMethod,
        ReturnFlag,
        OrderStatus,
        DaysToShip
    )
    SELECT
        DP.ProductKey,
        DC.CustomerKey,
        DD.DateKey,

        SS.OrderID,
        SS.Quantity,
        SS.DiscountPct,
        SS.SalesAmount,
        SS.Profit,
        SS.ShippingCost,
        SS.PaymentMethod,
        SS.ReturnFlag,
        SS.OrderStatus,
        SS.DaysToShip

    FROM silver.Sales_Silver SS

        INNER JOIN gold.DimProduct DP
            ON SS.SourceSalesKey = DP.SourceProductKey

        INNER JOIN gold.DimCustomer DC
            ON SS.CustomerID = DC.CustomerID

        INNER JOIN gold.DimDate DD
            ON SS.OrderDate = DD.FullDate;

END;
GO
```

# [etl].[usp_Load_Product_Silver]

```sql
/****** Object:  StoredProcedure [etl].[usp_Load_Product_Silver]    Script Date: 7/8/2026 9:44:10 PM ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO
CREATE PROCEDURE [etl].[usp_Load_Product_Silver]
AS
BEGIN
    SET NOCOUNT ON;

    TRUNCATE TABLE silver.Product_Silver;

    INSERT INTO silver.Product_Silver
    (
        ProductCategory,
        ProductName,
        UnitPrice,
        SourceProductID
    )
    SELECT
        ProductCategory,
        ProductName,
        TRY_CAST(NULLIF(LTRIM(RTRIM(UnitPrice)), '') AS DECIMAL(18,2)),
        ProductID
    FROM bronze.Product_Bronze
    WHERE ProductCategory IN
    (
        'Beauty',
        'Books',
        'Clothing',
        'Electronics',
        'Furniture',
        'Groceries',
        'Sports'
    );

END;
GO
```

# [etl].[usp_Load_Sales_Silver]

```sql
/****** Object:  StoredProcedure [etl].[usp_Load_Sales_Silver]    Script Date: 7/8/2026 9:44:10 PM ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO
CREATE PROCEDURE [etl].[usp_Load_Sales_Silver]
AS
BEGIN
    SET NOCOUNT ON;

    TRUNCATE TABLE silver.Sales_Silver;

    INSERT INTO silver.Sales_Silver
    (
        OrderID,
        OrderDate,
        CustomerID,
        Quantity,
        DiscountPct,
        SalesAmount,
        Profit,
        ShippingCost,
        PaymentMethod,
        ReturnFlag,
        OrderStatus,
        DaysToShip,
        SourceSalesKey
    )
    SELECT
        OrderID,
        TRY_CAST(OrderDate AS DATE) AS OrderDate,
        CustomerID,
        CAST(TRY_CAST(Quantity AS DECIMAL(10,2)) AS INT) AS Quantity,
        TRY_CAST(DiscountPct AS DECIMAL(5,2)) AS DiscountPct,
        TRY_CAST(SalesAmount AS DECIMAL(18,2)) AS SalesAmount,
        TRY_CAST(Profit AS DECIMAL(18,2)) AS Profit,
        TRY_CAST(ShippingCost AS DECIMAL(18,2)) AS ShippingCost,
        PaymentMethod,
        CASE
            WHEN ReturnFlag IN ('1','True','TRUE','Yes','YES')
                THEN 1
            ELSE 0
        END AS ReturnFlag,
        OrderStatus,
        CAST(TRY_CAST(DaysToShip AS DECIMAL(10,2)) AS INT) AS DaysToShip,
        SalesKey AS SourceSalesKey
    FROM bronze.Sales_Bronze;
END;
GO
```
